```python
import torch
import torch.nn as nn
```

```python
# input
torch.manual_seed(42) # similar to random_state in ML
inputs = torch.tensor([1.0,2.0,3.0])  # should be float values
```

```python
neuron = nn.Linear(in_features=3,out_features=1)
```

```python
output = neuron(inputs)
```

```python
output
```

```python
# verify
neuron.weight @ inputs + neuron.bias
```
